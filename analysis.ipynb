```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('survey_results_public.csv')
```

```python
#top 10 languages used 
langs = df['LanguageHaveWorkedWith'].dropna()
lang_list = langs.str.split(';').explode()
top_langs = lang_list.value_counts().head(10)

plt.figure(figsize=(10,6))
top_langs.plot(kind='bar', color='steelblue')
plt.title('Top 10 Programming Languages (SO Survey 2024)')
plt.tight_layout()
plt.savefig('charts/top_languages.png')
```

```python
#IA tools
ai_col = 'AISearchDevHaveWorkedWith'  # check exact column name
if ai_col in df.columns:
    ai = df[ai_col].dropna().str.split(';').explode()
    ai.value_counts().head(8).plot(kind='barh', color='coral')
    plt.title('Most Used AI Tools by Developers')
    plt.tight_layout()
    plt.savefig('charts/ai_tools.png')
```

```python
#remote 
remote = df['RemoteWork'].value_counts()
remote.plot(kind='pie', autopct='%1.1f%%', figsize=(7,7))
plt.title('Remote Work Distribution')
plt.savefig('charts/remote_work.png')
```

```python
#YEARS
df['YearsCode'].replace('More than 50 years', 51, inplace=True)
df['YearsCode'].replace('Less than 1 year', 0, inplace=True)
df['YearsCode'] = pd.to_numeric(df['YearsCode'], errors='coerce')
df['YearsCode'].dropna().plot(kind='hist', bins=20, color='teal')
plt.title('Developer Experience Distribution')
plt.xlabel('Years of Coding')
plt.savefig('charts/experience.png')
```

```python
#COUNTRIES
df['Country'].value_counts().head(15).plot(kind='bar', color='purple')
plt.title('Top 15 Countries by Developer Respondents')
plt.tight_layout()
plt.savefig('charts/countries.png')
```
